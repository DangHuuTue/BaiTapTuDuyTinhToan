```python
#Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):
def binary_search(arr, left, right, target):
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
arr = list(map(int, input().split()))
target = int(input())
result = binary_search(arr, 0, len(arr) - 1, target)
if result != -1:
    print(f"Phần tử {target} được tìm thấy tại chỉ số: {result}")
else:
    print(f"Phần tử {target} không được tìm thấy trong danh sách.")
```

```python
#Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) -> int:
def count_occurrences(arr, x) -> int:
    count = 0
    for num in arr:
        if num == x:
            count += 1
    return count
arr = list(map(int, input().split()))
x = int(input())
result = count_occurrences(arr, x)
print(f"Số lần xuất hiện của {x} trong danh sách là: {result}")
```

```python
#Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection Sort)
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
arr = list(map(int, input().split()))
sorted_arr_bubble = bubble_sort(arr.copy())
sorted_arr_selection = selection_sort(arr.copy())
print(f"Danh sách sau khi sắp xếp bằng Bubble Sort: {sorted_arr_bubble}")
print(f"Danh sách sau khi sắp xếp bằng Selection Sort: {sorted_arr_selection}")
```

```python
#Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.
def most_frequent(arr):
    frequency = {}
    max_count = 0
    most_frequent_element = arr[0]
    for num in arr:
        frequency[num] = frequency.get(num, 0) + 1
        if frequency[num] > max_count:
            max_count = frequency[num]
            most_frequent_element = num
    return most_frequent_element
arr = list(map(int, input().split()))
result = most_frequent(arr)
print(f"Phần tử xuất hiện nhiều nhất, sớm nhất trong danh sách là: {result}")
```

```python
#Cho danh sách các số nguyên và một số X, đếm tất cả các cặp (a, b) sao cho a + b = X.
def count_pairs(arr, x):
    count = 0
    seen = set()
    for num in arr:
        complement = x - num
        if complement in seen:
            count += 1
        seen.add(num)
    return count
arr = list(map(int, input().split()))
x = int(input())
result = count_pairs(arr, x)
print(f"Số cặp (a, b) sao cho a + b = {x} là: {result}")
```

```python
#Bạn được cung cấp một danh sách các danh sách lồng nhau, vd [[1, 2, 3], [4, 5], [6, 7, 8, 9]], hãy làm phẳng danh sách
def flatten(nested_list):
    flat_list = []
    for sublist in nested_list:
        for item in sublist:
            flat_list.append(item)
    return flat_list
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
result = flatten(nested_list)
print(f"Danh sách sau khi được làm phẳng: {result}")
```

```python
#Cho một danh sách các số nguyên nums, hãy tìm độ dài của dãy con tăng nghiêm ngặt dài nhất (không nhất thiết phải liền kề).
def longest_increasing_subsequence(nums):
    if not nums:
        return 0
    n = len(nums)
    dp = [1] * n
    for i in range(1, n):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)
nums = list(map(int, input().split()))
result = longest_increasing_subsequence(nums)
print(f"Độ dài của dãy con tăng nghiêm ngặt dài nhất là: {result}") 
```

```python
#Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách các truy vấn queries = [q1, q2, ...]  Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri.  Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list 
def min_interval(intervals, queries):
    intervals.sort(key=lambda x: x[1] - x[0]) 
    results = []
    for q in queries:
        min_length = float('inf')
        for l, r in intervals:
            if l <= q <= r:
                min_length = min(min_length, r - l + 1)
            elif l > q:  
                break
        results.append(min_length if min_length != float('inf') else -1)
    return results
intervals = [[1, 4], [2, 5], [3, 6]]
queries = [2, 3, 7]
result = min_interval(intervals, queries)
print(f"Kết quả cho các truy vấn: {result}")
```

```python
#Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có bao nhiêu phần tử bên phải nhỏ hơn nums[i].  Trả về danh sách các số đếm. Efficient solution: using heap or binary search tree
def count_smaller(nums):
    result = []
    sorted_list = []
    for num in reversed(nums):
        index = binary_search_insert_position(sorted_list, num)
        result.append(index)
        sorted_list.insert(index, num)
    return result[::-1]
def binary_search_insert_position(arr, target):
    left, right = 0, len(arr)
    while left < right:
        mid = left + (right - left) // 2
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid
    return left
nums = list(map(int, input().split()))
result = count_smaller(nums)
print(f"Danh sách số đếm: {result}")
```

```python
#Cho một danh sách các số guyên nums và một số nguyên m, hãy tìm tổng mảng con lớn nhất theo modulo m.  Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể. Gợi ý: Duy trì tổng tiền tố (prefix sum) mod m và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh).
def max_subarray_modulo(nums, m):
    prefix_sums = []
    current_sum = 0
    max_modulo = 0
    for num in nums:
        current_sum = (current_sum + num) % m
        max_modulo = max(max_modulo, current_sum)
        index = binary_search_insert_position(prefix_sums, current_sum)
        if index < len(prefix_sums):
            max_modulo = max(max_modulo, (current_sum - prefix_sums[index] + m) % m)
        prefix_sums.insert(index, current_sum)
    return max_modulo
def binary_search_insert_position(arr, target):
    left, right = 0, len(arr)
    while left < right:
        mid = left + (right - left) // 2
        if arr[mid] < target:
            left = mid + 1
        else:
            right = mid
    return left
nums = list(map(int, input().split()))
m = int(input())
result = max_subarray_modulo(nums, m)
print(f"Tổng mảng con lớn nhất theo modulo {m} là: {result}")
```

```python
#Các trình duyệt web thường có một chức năng tìm kiếm một từ trong một trang web. Chẳng hạn khi bạn tìm kiếm từ "UET" trên trang UETCodeHub, trình duyệt sẽ hiện ra có một từ "UET". Hãy viết một chương trình để xác định xem có bao nhiêu từ xuất hiện trong một câu. Đầu vào: Một chuỗi kí tự trên một dòng. Từ khóa cần tìm ở dòng tiếp theo. Đầu ra: Số lần xuất hiện của từ tìm kiếm. 
def count_word_occurrences(sentence, keyword):
    words = sentence.split()
    count = sum(1 for word in words if word == keyword)
    return count
sentence = input()
keyword = input()
result = count_word_occurrences(sentence, keyword)
print(f"Số lần xuất hiện của {keyword} trong câu là: {result}")
```

```python
#Dãy hài hòa là một dãy thỏa mãn điều kiện giá trị lớn nhất và giá trị nhỏ nhất của dãy khác nhau đúng bằng \( 1 \). Viết hàm trả về độ dài của dãy hài hòa con dài nhất có thể có trong mảng giá trị nguyên \( arr \) có \( n \) phần tử Ví dụ: Input: 1 3 2 2 5 2 3 7 Output: 5 - Dãy hài hòa con dài nhất là [3,2,2,2,3] 
def longest_harmonious_subsequence(arr):
    from collections import Counter
    count = Counter(arr)
    max_length = 0
    for num in count:
        if num + 1 in count:
            max_length = max(max_length, count[num] + count[num + 1])
    return max_length
arr = list(map(int, input().split()))
result = longest_harmonious_subsequence(arr)
print(f"Độ dài của dãy hài hòa con dài nhất là: {result}")
```
