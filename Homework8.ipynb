```python
#Xây dựng lớp Shape với phương thức area(). Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area(). Lấy Pi = 3.14.
class Shape:
    def area(self): pass
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return 3.14 * (self.r ** 2)
t = input().strip()
if t == "RECTANGLE":
    w, h = map(float, input().split())
    print(f"{Rectangle(w, h).area():.2f}")
elif t == "CIRCLE":
    r = float(input())
    print(f"{Circle(r).area():.2f}")
```

```python
# W8A17: Xây dựng lớp Vehicle(brand, speed). Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle. Viết phương thức info() cho mỗi lớp: Vehicle: brand-speed, Car: brand-speed-seats.
class Vehicle:
    def __init__(self, b, s):
        self.b = b
        self.s = s
    def info(self):
        return f"{self.b}-{self.s}"
class Car(Vehicle):
    def __init__(self, b, s, e):
        super().__init__(b, s)
        self.e = e
    def info(self):
        return f"{self.b}-{self.s}-{self.e}"
t = input().strip()
p = input().split()
if t == "Vehicle":
    print(Vehicle(p[0], p[1]).info())
elif t == "Car":
    print(Car(p[0], p[1], p[2]).info())
```

```python
# W8A18: Xây dựng lớp Polygon với phương thức perimeter(). Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.
class Polygon:
    def perimeter(self): pass
class Triangle(Polygon):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def perimeter(self):
        return self.a + self.b + self.c
a, b, c = map(int, input().split())
print(Triangle(a, b, c).perimeter())
```

```python
# W8A19: Xây dựng lớp Person(name, age) và lớp Student(name, age, school) kế thừa từ Person. Phương thức describe(): Person: name age, Student: name age school.
class Person:
    def __init__(self, n, a):
        self.n = n
        self.a = a
    def describe(self):
        return f"{self.n} {self.a}"
class Student(Person):
    def __init__(self, n, a, s):
        super().__init__(n, a)
        self.s = s
    def describe(self):
        return f"{self.n} {self.a} {self.s}"
t = input().strip()
p = input().split()
if t == "Person":
    print(Person(p[0], p[1]).describe())
elif t == "Student":
    print(Student(p[0], p[1], p[2]).describe())
```

```python
# W8A20: Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance(). Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức add_interest() để cộng balance*rate vào số dư.
class Account:
    def __init__(self, b):
        self.b = b
    def deposit(self, x):
        self.b += x
    def get_balance(self):
        return self.b
class SavingAccount(Account):
    def __init__(self, b, r):
        super().__init__(b)
        self.r = r
    def add_interest(self):
        self.b += self.b * self.r
b, r = map(float, input().split())
s = SavingAccount(b, r)
s.add_interest()
print(f"{s.get_balance():.2f}")
```

```python
# W8A21: Xây dựng lớp Printer với phương thức print_value(x). Xây dựng lớp BinaryPrinter kế thừa từ Printer, ghi đè print_value(x) để in biểu diễn nhị phân của số nguyên dương x.
class Printer:
    def print_value(self, x): pass
class BinaryPrinter(Printer):
    def print_value(self, x):
        print(bin(x)[2:])
n = int(input())
BinaryPrinter().print_value(n)
```

```python
# W8A22: Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận. Viết phương thức trace() trả về a+d và determinant() trả về ad - bc.
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a * self.d - self.b * self.c
a, b, c, d = map(int, input().split())
m = Matrix2x2(a, b, c, d)
print(f"{m.trace()} {m.determinant()}")
```

```python
# W8A23: Xây dựng lớp Polynomial biểu diễn đa thức bậc nhất ax + b. Viết phương thức evaluate(x) để tính giá trị đa thức tại x.
class Polynomial:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def evaluate(self, x):
        return self.a * x + self.b
a, b, x = map(int, input().split())
print(Polynomial(a, b).evaluate(x))
```

```python
# W8A24: Xây dựng lớp Polynomial2 biểu diễn đa thức bậc hai ax^2 + bx + c. Viết phương thức evaluate(x) để tính giá trị đa thức tại x.
class Polynomial2:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def evaluate(self, x):
        return self.a * (x ** 2) + self.b * x + self.c
a, b, c, x = map(int, input().split())
print(Polynomial2(a, b, c).evaluate(x))
```

```python
# W8A25: Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất: P(x)=ax+b, Q(x)=cx+d. Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        if m >= 0:
            return f"{k}x + {m}"
        return f"{k}x - {abs(m)}"
a, b, c, d = map(int, input().split())
print(PolynomialSum(a, b, c, d).sum_str())
```

```python
# W8A27: Xây dựng lớp Product(name, price). Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức final_price() để trả về giá sau giảm.
class Product:
    def __init__(self, n, p):
        self.n = n
        self.p = p
class DiscountProduct(Product):
    def __init__(self, n, p, c):
        super().__init__(n, p)
        self.c = c
    def final_price(self):
        return self.p * (100 - self.c) / 100
n, p, c = input().split()
p, c = float(p), float(c)
print(f"{DiscountProduct(n, p, c).final_price():.2f}")
```

```python
# W8A28: Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), value(). Sau đó thực hiện dãy thao tác trên đối tượng.
class Counter:
    def __init__(self):
        self.c = 0
    def inc(self):
        self.c += 1
    def dec(self):
        self.c -= 1
    def value(self):
        return self.c
n = int(input())
o = input().split()
g = Counter()
for x in o:
    if x == "INC":
        g.inc()
    elif x == "DEC":
        g.dec()
print(g.value())
```

```python
# W8A29: Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s
h, m, s = map(int, input().split())
print(Time(h, m, s).to_seconds())
```

```python
# W8A30: Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x), pop(), top().
class Stack:
    def __init__(self):
        self.s = []
    def push(self, x):
        self.s.append(x)
    def pop(self):
        return self.s.pop()
    def top(self):
        return self.s[-1]
n = int(input())
k = Stack()
for _ in range(n):
    c = input().split()
    if c[0] == "PUSH":
        k.push(c[1])
    elif c[0] == "POP":
        k.pop()
    elif c[0] == "TOP":
        print(k.top())
```
