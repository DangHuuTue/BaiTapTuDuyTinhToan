```python
#Nhập vào hai số nguyên a, b. Thực hiện in ra kết quả phép chia a/b. Bắt lỗi ngoại lệ khi b bằng 0. Kết quả in ra chính xác tới 2 chữ số thập phân nếu b khác 0. (Dùng try ... except bắt lỗi ZeroDivisionError) 
a = int(input())
b = int(input())
try:
    result = a / b
    print(f"{result:.2f}")
except ZeroDivisionError:
    print("Không thể chia cho 0")
```

```python
# [Invalid_Index] Hãy nhập vào một mảng my_list (list) n số nguyên (n nguyên dương). Thực hiện thử lệnh print(my_list[n]) rồi đọc kỹ Traceback và bắt lỗi Truy cập chỉ số sai. (Dùng try... except bắt lỗi IndexError)
try:
    n = int(input())
    x = [int(i) for i in input().split()]
    print(x[n])
except IndexError:
    print("Truy cap chi so sai")
```

```python
# [ValueError - HappyNumbers] Nhập vào một số nguyên dương a, Kiểm tra xem a có phải là số hạnh phúc không. Một số nguyên dương n được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước sau: Tính tổng bình phương các chữ số của n. Lặp lại bước đó với kết quả vừa tính. Nếu cuối cùng kết quả là 1, thì n là số hạnh phúc. Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như 4 -> 16 -> 37 -> ...), thì n không phải là số hạnh phúc. Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try... except bắt lỗi ValueError thêm set)
try:
    a = int(input())
    if a <= 0:
        raise ValueError
    s = set()
    while a != 1 and a not in s:
        s.add(a)
        a = sum(int(i)**2 for i in str(a))
    if a == 1:
        print("La so hanh phuc")
    else:
        print("Khong phai so hanh phuc")
except ValueError:
    print("Loi nhap lieu")
```

```python
# Sử dụng try except, kết hợp cùng else... finally. Tính tổng 2 số, nếu nhập sai kiểu thì thông báo, luôn luôn in "Kết thúc chương trình." sau khi chạy.
try:
    a = float(input())
    b = float(input())
except ValueError:
    print("Nhập sai kiểu")
else:
    print(a + b)
finally:
    print("Kết thúc chương trình.")
```

```python
# Viết chương trình chuyển tất cả phần tử của một list cho trước về int. Nếu phần tử nào trong list không thể chuyển được thì in cảnh báo và bỏ qua phần tử đó
x = list(input().split())
rs = []
for i in x:
    try:
        rs.append(int(i))
    except ValueError:
        print(f"Không thể chuyển {i} sang int")
```

```python
# Viết chương trình nhập một số x và tính căn bậc hai của nó. Bắt lỗi ngoại lệ khi x là số âm. Kết quả in ra chính xác tới 2 chữ số thập phân nếu x hợp lệ. (Dùng try... except bắt lỗi ValueError in ra "So am")
try:
    x = float(input())
    if x < 0:
        raise ValueError
    print(f"{x**0.5:.2f}")
except ValueError:
    print("So am")
```

```python
# Viết chương trình nhập vào số tuổi của một người. Nếu số tuổi nhỏ hơn 0 hoặc không phải số, in ra "Tuoi khong hop le". Nếu số tuổi hợp lệ thì in ra năm sinh của người ta (tỉnh từ năm 2025 trở về).
try:
    t = int(input())
    if t < 0:
        raise ValueError
    print(2025 - t)
except ValueError:
    print("Tuoi khong hop le")
```

```python
# Viết chương trình nhập vào tên file và kiểm tra xem file có đúng định dạng txt hay zip không. Nếu file hợp lệ thì in ra "Doc file thanh cong". Nếu file không hợp lệ, in ra "File khong hop le".
try:
    f = input()
    if f.endswith(".txt") or f.endswith(".zip"):
        print("Doc file thanh cong")
    else:
        raise ValueError
except ValueError:
    print("File khong hop le")
```

```python
# Viết chương trình nhập vào một chuỗi và chuyển chuỗi đó thành số nguyên (Ví dụ nhập '123' chuyển thành 123). In ra số nguyên sau khi chuyển. Nếu không thể chuyển, in ra "Chuoi khong hop le".
try:
    s = input()
    n = int(s)
    print(n)
except ValueError:
    print("Chuoi khong hop le")
```

```python
# Viết chương trình nhập vào phần tử N (N>0) và một mảng gồm N số nguyên dương. Một mảng là hợp lệ nếu nó gồm các số khác nhau. In ra tổng các phần tử nếu mảng đó là hợp lệ. Nếu mảng không hợp lệ, in ra "Mang khong hop le".
try:
    n = int(input())
    if n <= 0:
        raise ValueError
    a = [int(i) for i in input().split()]
    if len(a) != n or any(i <= 0 for i in a) or len(set(a)) != len(a):
        raise ValueError
    print(sum(a))
except ValueError:
    print("Mang khong hop le")
```

```python
# [Parse_Float] Nhập vào một chuỗi s. Hãy chuyển s thành số thực và in ra giá trị làm tròn đến 2 chữ số thập phân. Nếu không thể chuyển, in ra "Du lieu khong hop le".
try:
    s = input()
    f = float(s)
    print(f"{f:.2f}")
except ValueError:
    print("Du lieu khong hop le")
```

```python
# [KeyError - StudentScore] Cho trước từ điển điểm gồm các cặp khóa - giá trị: {"An": 8, "Binh": 7, "Chi": 9}. Nhập vào tên học sinh và in ra điểm tương ứng. Nếu tên không tồn tại trong từ điển, bắt lỗi và in ra "Khong tim thay hoc sinh".
d = {"An": 8, "Binh": 7, "Chi": 9}
try:
    k = input()
    print(d[k])
except KeyError:
    print("Khong tim thay hoc sinh")
```

```python
# [IndexError - PopElement] Nhập vào một danh sách số nguyên và một chỉ số k. Hãy in ra phần tử bị xóa khi thực hiện pop(k). Nếu k không hợp lệ, in ra "Chi so khong hop le".
try:
    a = list(map(int, input().split()))
    k = int(input())
    print(a.pop(k))
except (IndexError, ValueError):
    print("Chi so khong hop le")
```

```python
# [TypeError - SumNumbers] Nhập vào một danh sách gồm nhiều giá trị. Hãy cộng tất cả phần tử sau khi chuyển được sang int. Nếu phần tử nào không chuyển được thì bỏ qua. In ra tổng cuối cùng.
a = input().split()
s = 0
for i in a:
    try:
        s += int(i)
    except ValueError:
        pass
print(s)
```

```python
# [Raise_Exception - Positive Number] Nhập vào một số nguyên n. Nếu n<=0 thì chủ động phát sinh ngoại lệ và in ra "So phai duong". Nếu hợp lệ thì in ra bình phương của n.
try:
    n = int(input())
    if n <= 0:
        raise Exception
    print(n**2)
except Exception:
    print("So phai duong")
```
