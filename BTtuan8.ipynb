```python
# Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần. Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.
def a():
    import collections
    b = list(map(int, input().split()))
    c = collections.Counter(b)
    y = 0
    for k, v in c.items():
        if k == 0:
            y += v
        elif v >= k:
            y += v - k
        else:
            y += v
    if y == 0:
        print("YES")
    else:
        print("NO", y)
a()
```

```python
# Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ phức tạp thời gian là O(n) độ phức tạp bộ nhớ là O(1), - Boyer-Moore. Nếu không có phần tử nào thỏa mãn thì không in gì
def b():
    a = list(map(int, input().split()))
    m = None
    c = 0
    for x in a:
        if c == 0:
            m = x
            c = 1
        elif x == m:
            c += 1
        else:
            c -= 1
    if a.count(m) > len(a) // 2:
        print(m)
b()
```

```python
# Sắp xếp có điều kiện: Số chẵn đứng trước số lẻ, đứng sau, chẵn tăng dần, lẻ giảm dần
def c():
    a = list(map(int, input().split()))
    e = sorted([x for x in a if x % 2 == 0])
    o = sorted([x for x in a if x % 2 != 0], reverse=True)
    print(*(e + o))
c()
```

```python
# Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào thành một mảng cũng có cùng thứ tự sắp xếp như mảng ban đầu
def d():
    x = list(map(int, input().split()))
    y = list(map(int, input().split()))
    if not x:
        print(*(y))
        return
    if not y:
        print(*(x))
        return
    a = (
        x[0] <= x[-1]
        if len(x) > 1
        else (y[0] <= y[-1] if len(y) > 1 else True)
    )
    r = []
    i = 0
    j = 0
    while i < len(x) and j < len(y):
        if a:
            if x[i] <= y[j]:
                r.append(x[i])
                i += 1
            else:
                r.append(y[j])
                j += 1
        else:
            if x[i] >= y[j]:
                r.append(x[i])
                i += 1
            else:
                r.append(y[j])
                j += 1
    while i < len(x):
        r.append(x[i])
        i += 1
    while j < len(y):
        r.append(y[j])
        j += 1
    print(*r)
d()
```

```python
# Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi N số nguyên dương, dòng sau ghi số nguyên dương k
def e():
    a = input().split()
    k = int(input())
    s = sum(int(x) for x in a if len(x) == k)
    print(s)
e()
```

```python
# Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu của năm đó nếu đó là ngày hợp lệ
def f():
    n, t = map(int, input().split())
    m = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if t < 1 or t > 12 or n < 1 or n > m[t - 1]:
        print("invalid")
    else:
        print(sum(m[: t - 1]) + n)
f()
```

```python
# Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất. Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất
def g():
    n = int(input())
    d = {}
    for i in range(n):
        s = input().split()[-1]
        d[s] = d.get(s, 0) + 1
    m = max(d.values())
    r = sorted([k for k, v in d.items() if v == m])[0]
    print(r)
g()
```

```python
# Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí tự cái.
def h():
    s = input()
    w = ""
    a = []
    for c in s:
        if c.isalpha():
            w += c
        else:
            if w:
                a.append(w)
            w = ""
    if w:
        a.append(w)
    r = 0
    for x in a:
        if len(x) >= 2 and x[0].isupper() and x[1:].islower():
            r += 1
    print(r)
h()
```

```python
# Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên
def i():
    a = input().split()
    r = []
    for x in a:
        c = "".join(e for e in x if e.isalnum()).lower()
        if not r:
            r.append(x)
        else:
            p = "".join(e for e in r[-1] if e.isalnum()).lower()
            if c != p:
                r.append(x)
    print(" ".join(r))
i()
```

```python
# Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số
def j():
    b = input().split()
    s = b[0]
    k = int(b[1]) if len(b) > 1 else int(input())
    m = len(s) - k
    g = []
    for d in s:
        while m > 0 and g and g[-1] < d:
            g.pop()
            m -= 1
        g.append(d)
    print("".join(g[:k]))
j()
```

```python
# Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó.
def k():
    s = input()
    v = ""
    a = []
    for c in s:
        if c.isdigit():
            v += c
        else:
            if v:
                a.append(int(v))
            v = ""
    if v:
        a.append(int(v))
    if not a:
        print("0 0")
        return
    g = sum(a)
    n = len(a)
    if g % n == 0:
        print(f"{g} {g // n}")
    else:
        print(f"{g} {g / n}")
k()
```

```python
# Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không
def l():
    s = input().strip()
    g = []
    d = {")": "(", "}": "{", "]": "["}
    v = True
    for c in s:
        if c in d.values():
            g.append(c)
        elif c in d:
            if g and g[-1] == d[c]:
                g.pop()
            else:
                v = False
                break
    if v and not g:
        print("YES")
    else:
        print("NO")
l()
```

```python
# Bài toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên nào <=N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)
def m():
    n = int(input())
    def p(x):
        if x < 2:
            return False
        for i in range(2, int(x**0.5) + 1):
            if x % i == 0:
                return False
        return True
    b = []
    c = []
    for x in range(2, n):
        if x > 1 and not p(x):
            c.append(x)
        else:
            if len(c) > len(b):
                b = c
            c = []
    if len(c) > len(b):
        b = c
    print(", ".join(map(str, b)))
m()
```

```python
# Tìm Dây con tăng dần liên tục dài nhất của một danh sách nhập vào từ bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau
def o():
    a = list(map(int, input().split()))
    if not a:
        return
    b = []
    c = [a[0]]
    for i in range(1, len(a)):
        if a[i] > a[i - 1]:
            c.append(a[i])
        else:
            if len(c) > len(b):
                b = c
            c = [a[i]]
    if len(c) > len(b):
        b = c
    print(*(b))
o()
```

```python
# Tìm Dây con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím
def p():
    a = list(map(int, input().split()))
    if not a:
        return
    n = len(a)
    d = [1] * n
    g = [-1] * n
    for i in range(n):
        for j in range(i):
            if a[i] > a[j] and d[j] + 1 > d[i]:
                d[i] = d[j] + 1
                g[i] = j
    m = max(d)
    k = d.index(m)
    r = []
    while k != -1:
        r.append(a[k])
        k = g[k]
    print(*(r[::-1]))
p()
```

```python
# Đổi chỗ đúng 2 chữ số để được số nhỏ nhất có thể
def q():
    s = input()
    a = list(s)
    n = len(a)
    i = 0
    while i < n - 1 and a[i] <= a[i + 1]:
        i += 1
    if i == n - 1:
        print(s)
        return
    j = n - 1
    while j > i and a[j] >= a[i]:
        j -= 1
    a[i], a[j] = a[j], a[i]
    print("".join(a))
q()
```
