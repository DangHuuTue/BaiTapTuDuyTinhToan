```python
# Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần. Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.
from collections import Counter
def f(a):
    c = Counter(a)
    y = 0
    for x, v in c.items():
        if x == 0:
            y += v
        elif v >= x:
            y += v - x
        else:
            y += v
    return y
a = list(map(int, input().split()))
print(f(a))
```

```python
# Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer-Moore. Nếu không có phần tử nào thỏa mãn thì không in gì
def majority_element(nums):
    count = 0
    candidate = None
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    if nums.count(candidate) > len(nums) // 2:
        return candidate
    else:
        return None
nums = list(map(int, input().split()))
result = majority_element(nums)
if result is not None:
    print(result)
```

```python
# Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần
def f(a):
    b = sorted([x for x in a if x % 2 == 0])
    c = sorted([x for x in a if x % 2 != 0], reverse=True)
    return b + c
a = list(map(int, input().split()))
print(f(a))
```

```python
# Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào
def f(a, b):
    if len(a) > 1 and a[0] > a[-1]:
        a = a[::-1]
    if len(b) > 1 and b[0] > b[-1]:
        b = b[::-1]
    r = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            r.append(a[i])
            i += 1
        else:
            r.append(b[j])
            j += 1
    r.extend(a[i:])
    r.extend(b[j:])
    return r
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(f(a, b))
```

```python
# Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu của năm đó nếu đó là ngảy hợp lệ
def f(n, t):
    m = [0, 31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if t < 1 or t > 12 or n < 1 or n > m[t]:
        return "Ngay khong hop le"
    return sum(m[:t]) + n
n = int(input())
t = int(input())
print(f(n, t))
```

```python
# Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất. Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất
def f(names):
    c = Counter(name.split()[-1] for name in names)
    max_count = max(c.values())
    most_common_names = [name for name, count in c.items() if count == max_count]
    return min(most_common_names)
names = [input() for _ in range(int(input()))]
print(f(names))
```

```python
# Đếm số lượng tên trong xâu kí tự. Tên được định nghĩa là xâu có độ dài >=2 và có chữ cái đầu viết hoa, các chữ cái khác viết thường, chỉ có kí tự cái.
import re
def f(s):
    w = re.findall(r"\b[A-Z][a-z]+\b", s)
    return len(w)
s = input()
print(f(s))
```

```python
# Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên
def f(s):
    w = s.split()
    if not w:
        return ""
    r = [w[0]]
    for i in range(1, len(w)):
        if w[i] != w[i - 1]:
            r.append(w[i])
    return " ".join(r)
s = input()
print(f(s))
```

```python
# Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số
def f(s, k):
    r = []
    r = []
    n = len(s)
    for i, d in enumerate(s):
        while r and r[-1] < d and len(r) + n - i > k:
            r.pop()
        if len(r) < k:
            r.append(d)
    return "".join(r)
s = input()
k = int(input())
print(f(s, k))
```

```python
# Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các số đó.
def f(s):
    numbers = re.findall(r'-?\d+', s)
    numbers = [int(x) for x in numbers]
    total = sum(numbers)
    average = total / len(numbers) if numbers else 0
    return total, average
s = input()
total, average = f(s)
print(f"Tổng: {total}, Trung bình: {average}")
```

```python
# Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không
def f(s):
    d = {"(": ")", "{": "}", "[": "]"}
    r = []
    for c in s:
        if c in d:
            r.append(c)
        elif r and d[r[-1]] == c:
            r.pop()
        else:
            return False
    return len(r) == 0
s = input()
print(f(s))
```

```python
# Bài toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên nào <=N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)
def f(n):
    def p(x):
        if x < 2:
            return False
        for i in range(2, int(x**0.5) + 1):
            if x % i == 0:
                return False
        return True
    i = n + 1
    while True:
        if p(i):
            a = i
            break
        i += 1
    while True:
        i += 1
        if p(i):
            b = i
            break
    return list(range(a + 1, b))
n = int(input())
print(f(n))
```

```python
# Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau
def f(a):
    if not a:
        return []
    m = []
    c = [a[0]]
    for i in range(1, len(a)):
        if a[i] > a[i - 1]:
            c.append(a[i])
        else:
            if len(c) > len(m):
                m = c
            c = [a[i]]
    if len(c) > len(m):
        m = c
    return m
a = list(map(int, input().split()))
print(f(a))
```

```python
# Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập vào từ bàn phím
def f(a):
    if not a:
        return []
    n = len(a)
    f = [1] * n
    p = [-1] * n
    for i in range(1, n):
        for j in range(i):
            if a[i] > a[j] and f[j] + 1 > f[i]:
                f[i] = f[j] + 1
                p[i] = j
    m = max(f)
    i = f.index(m)
    r = []
    while i != -1:
        r.append(a[i])
        i = p[i]
    return r[::-1]
a = list(map(int, input().split()))
print(f(a))
```
