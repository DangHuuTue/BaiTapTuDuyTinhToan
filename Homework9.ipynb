```python
# Cho hai số nguyên dương, n và m (0 < n, m <= 20). Bạn hãy in ra tất cả các chuỗi nhị phân có n số 0 và m số 1. In theo thứ tự tăng dần trong từ điển
n = int(input())
m = int(input())
def f(n, m, s):
    if n == 0 and m == 0:
        print(s)
        return
    if n > 0:
        f(n - 1, m, s + "0")
    if m > 0:
        f(n, m - 1, s + "1")
f(n, m, "")
```

```python
# Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím
def f():
    a = list(map(int, input().split()))
    if not a:
        return
    v = [1] * len(a)
    for i in range(len(a)):
        for j in range(i):
            if a[i] > a[j]:
                v[i] = max(v[i], v[j] + 1)
    print(max(v))
f()
```

```python
# Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên dương cho vào từ bàn phím)
a = list(map(int, input().split()))
k = int(input())
c = 0
for i in range(len(a)):
    for j in range(i + 1, len(a)):
        if a[i] + a[j] == k:
            c += 1
print(c)
```

```python
# Loại bỏ khỏi danh sách các phần tử trùng lặp, giữ lại đúng 1 lần xuất hiện, in lại theo đúng thứ tự xuất hiện
a = input().split()
b = []
s = set()
for x in a:
    if x not in s:
        b.append(x)
        s.add(x)
print(*(b))
```

```python
# Thực hiện chuẩn hóa chuỗi kí tự:
# a. bỏ khoảng trắng đầu cuối
# b. các khoảng trắng giữa chỉ còn 1 dấu cách
# c. viết hoa chữ cái đầu mỗi từ
# KHÔNG DÙNG CÁC HÀM SẴN CÓ
s = input()
i = 0
while i < len(s) and s[i] == ' ':
    i += 1
j = len(s) - 1
while j >= i and s[j] == ' ':
    j -= 1

r = ""
f = True
for k in range(i, j + 1):
    if s[k] == ' ':
        if not f:
            r += ' '
            f = True
    else:
        c = s[k]
        if f:
            if 'a' <= c <= 'z':
                c = chr(ord(c) - 32)
            f = False
        else:
            if 'A' <= c <= 'Z':
                c = chr(ord(c) + 32)
        r += c
print(r)
```

```python
# Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp (dùng phân hoạch của quicksort)
def q(v, k):
    p = v[len(v) // 2]
    l = [x for x in v if x > p]
    m = [x for x in v if x == p]
    r = [x for x in v if x < p]
    if k <= len(l):
        return q(l, k)
    elif k <= len(l) + len(m):
        return p
    else:
        return q(r, k - len(l) - len(m))

a = list(map(int, input().split()))
k = int(input())
print(q(a, k))
```

```python
# Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi đầu vào có phải là anagram của nhau không
s = input()
t = input()
d = {}
for x in s:
    d[x] = d.get(x, 0) + 1
for x in t:
    d[x] = d.get(x, 0) - 1
f = True
for k in d:
    if d[k] != 0:
        f = False
print("YES" if f else "NO")
```

```python
# Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm)
s = input()
k = int(input())
d = len(s) - k
t = []
for x in s:
    while d > 0 and t and t[-1] < x:
        t.pop()
        d -= 1
    t.append(x)
print("".join(t[:k]))
```

```python
# Nén chuỗi kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau kí tự đó. VD aaaaaafdhhhhh => a6f1d1h5
s = input()
if s:
    r = ""
    c = 1
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            c += 1
        else:
            r += s[i - 1] + str(c)
            c = 1
    r += s[-1] + str(c)
    print(r)
```

```python
# Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem nó có phải là mật khẩu mạnh không.
s = input()
if len(s) >= 8:
    a = b = c = d = False
    for x in s:
        if 'a' <= x <= 'z': a = True
        elif 'A' <= x <= 'Z': b = True
        elif '0' <= x <= '9': c = True
        elif x in "!@#$": d = True
    if a and b and c and d:
        print("Mạnh")
    else:
        print("Không mạnh")
else:
    print("Không mạnh")
```
