```python
# W9A1: Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ thuật toán sắp xếp nào
def f(a, b):
    c = []
    i = 0
    j = 0
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            c.append(a[i])
            i += 1
        else:
            c.append(b[j])
            j += 1
    while i < len(a):
        c.append(a[i])
        i += 1
    while j < len(b):
        c.append(b[j])
        j += 1
    return c
a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(f(a, b))
```

```python
# W9A2: Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên dòng tăng dần và trên cột tăng dần, hãy tìm số lớn thứ k (k < nxm) với thuật toán tốt nhất có thể
def f(m, k):
    r = len(m)
    c = len(m[0])
    l = m[0][0]
    h = m[r-1][c-1]
    a = l
    while l <= h:
        g = (l + h) // 2
        i = 0
        j = c - 1
        s = 0
        while i < r and j >= 0:
            if m[i][j] >= g:
                s += r - i
                j -= 1
            else:
                i += 1
        if s >= k:
            a = g
            l = g + 1
        else:
            h = g - 1
    return a

r, c, k = map(int, input().split())
m = [list(map(int, input().split())) for x in range(r)]
print(f(m, k))
```

```python
# W9A3: Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ số để được một số nguyên nhỏ nhất có thể
def f(s):
    v = list(s)
    n = len(v)
    m = None
    for i in range(n):
        for j in range(i + 1, n):
            if i == 0 and v[j] == '0':
                continue
            v[i], v[j] = v[j], v[i]
            x = int("".join(v))
            if m is None or x < m:
                m = x
            v[i], v[j] = v[j], v[i]
    if m is None:
        for i in range(n):
            for j in range(i + 1, n):
                v[i], v[j] = v[j], v[i]
                x = int("".join(v))
                if m is None or x < m:
                    m = x
                v[i], v[j] = v[j], v[i]
    return m if m is not None else int(s)

s = input().strip()
print(f(s))
```

```python
# W9A4: Cho 2 xâu a,b, xác định xem có cách biến đổi từ xâu a sang xâu b không. Ghi ra "YES" nếu có và "NO" nếu ngược lại.
def f(a, b):
    n = len(a)
    m = len(b)
    d = [[False] * (m + 1) for x in range(n + 1)]
    d[0][0] = True
    for i in range(1, n + 1):
        if a[i-1].islower():
            d[i][0] = d[i-1][0]
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            if a[i-1].isupper():
                if a[i-1] == b[j-1]:
                    d[i][j] = d[i-1][j-1]
            else:
                d[i][j] = d[i-1][j]
                if a[i-1].upper() == b[j-1]:
                    d[i][j] = d[i][j] or d[i-1][j-1]
    return "YES" if d[n][m] else "NO"

q = int(input())
for x in range(q):
    a = input().strip()
    b = input().strip()
    print(f(a, b))
```

```python
# W9A5: Cho hai mảng số nguyên list1 và list2, viết hàm kiểm tra xem list1 có thuộc list2 hay không (có phải là dãy con liên tiếp của list2 hay không)?
def f(x, y):
    if not x:
        return True
    n = len(x)
    m = len(y)
    for i in range(m - n + 1):
        if y[i:i+n] == x:
            return True
    return False
x = list(map(int, input().split()))
y = list(map(int, input().split()))
print(f(x, y))
```

```python
# W9A6: [Hoán vị tiếp theo] Cho một dãy số gồm n chữ số là một hoán vị của dây 1,2,...,n. Viết chương trình in ra hoán vị ở vị trí tiếp theo.
def f(a):
    n = len(a)
    i = n - 2
    while i >= 0 and a[i] >= a[i+1]:
        i -= 1
    if i >= 0:
        j = n - 1
        while a[j] <= a[i]:
            j -= 1
        a[i], a[j] = a[j], a[i]
    l = i + 1
    r = n - 1
    while l < r:
        a[l], a[r] = a[r], a[l]
        l += 1
        r -= 1
    return a

n = int(input())
a = list(map(int, input().split()))
print(*(f(a)))
```

```python
# W9A7: [Can Place Flowers] Viết hàm canPlaceFlowers(flowerbed, k) kiểm tra xem có thể trồng hết số hoa mới được cấp vào trong luống đất của bạn hay không.
def f(b, k):
    n = len(b)
    for i in range(n):
        if b[i] == 0:
            l = (i == 0 or b[i-1] == 0)
            r = (i == n - 1 or b[i+1] == 0)
            if l and r:
                b[i] = 1
                k -= 1
                if k <= 0:
                    return True
    return k <= 0

b = list(map(int, input().split()))
k = int(input())
print(f(b, k))
```

```python
# W9A8: [Phát hiện đường đi lặp lại của Rô-bốt] Hàm trả về True nếu trong quá trình di chuyển, rô-bốt đi vào một ô mà nó đã từng đi qua trước đó.
def f(m):
    s = {(0, 0)}
    x, y = 0, 0
    for c in m:
        if c == 'R': x += 1
        elif c == 'L': x -= 1
        elif c == 'U': y += 1
        elif c == 'D': y -= 1
        if (x, y) in s:
            return True
        s.add((x, y))
    return False

m = input().strip()
print(f(m))
```

```python
# W9A9: [HARD - Tìm trung vị của 5 số] Tìm và trả về số trung vị (số lớn thứ ba) của 5 số này.
def f(v):
    a, b, c, d, e = v[0], v[1], v[2], v[3], v[4]
    if a > b: a, b = b, a
    if c > d: c, d = d, c
    if a > c: a, b, c, d = c, d, a, b
    if e < c:
        if b < c: return b if b > e else e
        else: return c
    else:
        if b < d:
            return b if b > e else (e if e < d else d)
        else:
            return d if e > d else e

v = list(map(int, input().split()))
print(f(v))
```

```python
# W9A10: [Tìm bán kính lò sưởi nhỏ nhất] Tìm bản kính r nhỏ nhất cho các lò sưởi sao cho mọi ngôi nhà đều nằm trong phạm vi sưởi ấm của ít nhất một lò sưởi.
def f(a, b):
    a.sort()
    b.sort()
    r = 0
    i = 0
    for x in a:
        while i < len(b) - 1 and abs(b[i+1] - x) <= abs(b[i] - x):
            i += 1
        r = max(r, abs(b[i] - x))
    return r

a = list(map(int, input().split()))
b = list(map(int, input().split()))
print(f(a, b))
```

```python
# W9A11: Viết hàm factorial(n) tính giai thừa của n bằng đệ quy.
def f(n):
    if n <= 1:
        return 1
    return n * f(n - 1)

n = int(input())
print(f(n))
```

```python
# W9A12: Viết hàm sum_to_n(n) tính tổng từ 1 đến n bằng đệ quy.
def f(n):
    if n <= 1:
        return n
    return n + f(n - 1)
n = int(input())
print(f(n))
```

```python
# W9A13: Viết hàm power(base, exp) tinh base^exp bằng đệ quy chia để trị.
def f(b, e):
    if e == 0:
        return 1
    t = f(b, e // 2)
    if e % 2 == 0:
        return t * t
    else:
        return b * t * t

b, e = map(int, input().split())
print(f(b, e))
```

```python
# W9A14: Viết hàm tìm số xuất hiện nhiều hơn x lần trong một dãy số (x là số nguyên dương nhập từ bàn phím). Nếu không có số nào thì in ra "INVALID"
def f(a, x):
    d = {}
    for i in a:
        d[i] = d.get(i, 0) + 1
    r = [k for k, v in d.items() if v > x]
    if not r:
        return "INVALID"
    return " ".join(map(str, r))

a = list(map(int, input().split()))
x = int(input())
print(f(a, x))
```

```python
# W9A15: Viết haàm kiểm tra dãy số nguyên vào từ bàn phím có phải dây tăng nghiêm ngặt (giảm nghiêm ngặt) hay không?
def f(a):
    if len(a) <= 1:
        return True
    i = all(a[x] < a[x+1] for x in range(len(a)-1))
    d = all(a[x] > a[x+1] for x in range(len(a)-1))
    return i or d
a = list(map(int, input().split()))
print(f(a))
```
